```python

def unify(x, y, substitution):

    if x == y:
        return True

    if x.islower():
        substitution[x] = y
        return True

    if y.islower():
        substitution[y] = x
        return True

    return False


# Get expressions from user
exp1 = input("Enter first expression: ")
exp2 = input("Enter second expression: ")

# Extract arguments
a = exp1[exp1.index("(") + 1 : -1].split(",")
b = exp2[exp2.index("(") + 1 : -1].split(",")

# Check predicates and number of arguments
p1 = exp1.split("(")[0]
p2 = exp2.split("(")[0]

substitution = {}

if p1 != p2 or len(a) != len(b):
    print("Unification failed")

else:
    success = True

    for x, y in zip(a, b):
        x = x.strip()
        y = y.strip()

        if not unify(x, y, substitution):
            success = False
            break

    if success:
        print("Unification successful")
        print("Substitution:", substitution)
    else:
        print("Unification failed")
```
